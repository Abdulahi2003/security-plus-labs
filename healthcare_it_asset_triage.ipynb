# Healthcare IT Asset Triage

This notebook is a practice exercise for reviewing a fictional device inventory. It demonstrates Markdown, Python code cells, output, simple filtering, and an AI-assisted learning workflow.

```python
devices = [
    {"name": "ER-WS-12", "type": "Windows workstation", "os": "Windows 11", "days_since_patch": 12, "encrypted": True},
    {"name": "NURSE-TAB-04", "type": "Tablet", "os": "Android", "days_since_patch": 47, "encrypted": True},
    {"name": "LAB-WS-03", "type": "Windows workstation", "os": "Windows 10", "days_since_patch": 95, "encrypted": False},
    {"name": "IMAGING-01", "type": "Imaging console", "os": "Vendor appliance", "days_since_patch": 160, "encrypted": True}
]

print(f"Loaded {len(devices)} fictional devices.")
```

```python
for device in devices:
    print(device["name"], "-", device["os"], "-", device["days_since_patch"], "days since patch")
```

```python
needs_attention = []

for device in devices:
    if device["days_since_patch"] > 30 or device["encrypted"] == False:
        needs_attention.append(device)

print(f"Devices needing review: {len(needs_attention)}")

for device in needs_attention:
    print("-", device["name"])
```

## What I Changed

I changed the patch review threshold from 60 days to 30 days. This caused devices with more than 30 days since their last patch to be flagged for review.

## What I Learned From an Error

I accidentally changed `devices` to `device` in the for loop. This caused a TypeError because the loop was working with strings instead of the device dictionaries I expected. I fixed the error by changing it back to `for device in devices:`.
